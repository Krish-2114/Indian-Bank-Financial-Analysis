```python
# ==========================================
# LIABILITY DATA — CLEANING & STANDARDIZATION
# ==========================================

# Load both liability sheets
liab_old = pd.read_excel(
    file_path,
    sheet_name="LIABILITIES_Upto2019",
    header=6
)

liab_new = pd.read_excel(
    file_path,
    sheet_name="LIABILITIES_2020_Onwards",
    header=6
)

# Remove blank first column
liab_old = liab_old.drop(columns=["Unnamed: 0"])
liab_new = liab_new.drop(columns=["Unnamed: 0"])

# Fill down Year
liab_old["Year"] = liab_old["Year"].ffill()
liab_new["Year"] = liab_new["Year"].ffill()

# Convert Year to numeric
liab_old["Year"] = pd.to_numeric(liab_old["Year"], errors="coerce")
liab_new["Year"] = pd.to_numeric(liab_new["Year"], errors="coerce")

# Remove non-data rows
liab_old = liab_old.dropna(subset=["Year"])
liab_new = liab_new.dropna(subset=["Year"])

# Convert Year to integer
liab_old["Year"] = liab_old["Year"].astype(int)
liab_new["Year"] = liab_new["Year"].astype(int)

# Remove aggregate rows
aggregate_names = [
    "PUBLIC SECTOR BANKS",
    "PRIVATE SECTOR BANKS",
    "FOREIGN BANKS",
    "SMALL FINANCE BANKS",
    "ALL SCHEDULED COMMERCIAL BANKS"
]

liab_old = liab_old[
    ~liab_old["Banks"].isin(aggregate_names)
].copy()

liab_new = liab_new[
    ~liab_new["Banks"].isin(aggregate_names)
].copy()


# ==========================================
# SELECT CORE LIABILITY COLUMNS
# ==========================================

liab_old_core = liab_old[
    [
        "Year",
        "Banks",
        "1.     Capital",
        "2.     Reserves and Surplus",
        "3.     Deposits",
        "4.     Borrowings",
        "5.     Other liabilities & provisions",
        "Total Liabilities"
    ]
].copy()


liab_new_core = liab_new[
    [
        "Year",
        "Banks",
        "1.     Capital",
        "2. Reserves and Surplus",
        "3. Deposits",
        "4. Borrowings",
        "5. Other liabilities & provisions",
        "Total Liabilities"
    ]
].copy()


# ==========================================
# STANDARDIZE COLUMN NAMES
# ==========================================

liab_old_core = liab_old_core.rename(columns={
    "2.     Reserves and Surplus": "Reserves and Surplus",
    "3.     Deposits": "Deposits",
    "4.     Borrowings": "Borrowings",
    "5.     Other liabilities & provisions": "Other liabilities & provisions"
})

liab_new_core = liab_new_core.rename(columns={
    "2. Reserves and Surplus": "Reserves and Surplus",
    "3. Deposits": "Deposits",
    "4. Borrowings": "Borrowings",
    "5. Other liabilities & provisions": "Other liabilities & provisions"
})


# ==========================================
# CONVERT FINANCIAL VALUES TO NUMERIC
# ==========================================

liability_columns = [
    "1.     Capital",
    "Reserves and Surplus",
    "Deposits",
    "Borrowings",
    "Other liabilities & provisions",
    "Total Liabilities"
]

for col in liability_columns:
    liab_old_core[col] = pd.to_numeric(
        liab_old_core[col],
        errors="coerce"
    )

    liab_new_core[col] = pd.to_numeric(
        liab_new_core[col],
        errors="coerce"
    )


# ==========================================
# COMBINE BOTH PERIODS
# ==========================================

liabilities_core = pd.concat(
    [liab_old_core, liab_new_core],
    ignore_index=True
)


# ==========================================
# CLEAN IDENTIFIERS
# ==========================================

liabilities_core["Banks"] = (
    liabilities_core["Banks"]
    .astype(str)
    .str.strip()
)

liabilities_core = liabilities_core.dropna(
    subset=["Year", "Banks"]
)


# ==========================================
# CHECK RESULT
# ==========================================

print("Liability dataset shape:", liabilities_core.shape)

display(liabilities_core.head())

print("\nYears covered:")
print(
    liabilities_core["Year"]
    .min(),
    "to",
    liabilities_core["Year"].max()
)

print("\nMissing values:")
print(liabilities_core.isna().sum())
```

```python
# ==========================================
# PREPARE ASSETS + LIABILITIES FOR MERGE
# ==========================================

# Remove rows with missing identifiers
assets_core = assets_core.dropna(
    subset=["Year", "Banks"]
).copy()

liabilities_core = liabilities_core.dropna(
    subset=["Year", "Banks"]
).copy()

# Standardize bank names
assets_core["Banks"] = (
    assets_core["Banks"]
    .astype(str)
    .str.strip()
)

liabilities_core["Banks"] = (
    liabilities_core["Banks"]
    .astype(str)
    .str.strip()
)

# Check for duplicate Bank-Year combinations
print("Asset duplicates:",
      assets_core.duplicated(["Year", "Banks"]).sum())

print("Liability duplicates:",
      liabilities_core.duplicated(["Year", "Banks"]).sum())
```

```python
# ==========================================
# MERGE ASSETS + LIABILITIES
# ==========================================

financials = assets_core.merge(
    liabilities_core,
    on=["Year", "Banks"],
    how="inner",
    validate="one_to_one"
)

print("Merged dataset shape:", financials.shape)

display(financials.head())

print("\nYears covered:")
print(financials["Year"].min(), "to", financials["Year"].max())

print("\nUnique banks:")
print(financials["Banks"].nunique())

print("\nMissing values:")
print(financials.isna().sum())
```

```python
# ==========================================
# RENAME FINANCIAL COLUMNS
# ==========================================

financials = financials.rename(columns={
    "Year": "year",
    "Banks": "bank_name",
    "6.     Investments": "investments",
    "7.     Advances": "advances",
    "8.     Fixed Assets": "fixed_assets",
    "9.     Other Assets": "other_assets",
    "Total Assets": "total_assets",
    "1.     Capital": "capital",
    "Reserves and Surplus": "reserves_surplus",
    "Deposits": "deposits",
    "Borrowings": "borrowings",
    "Other liabilities & provisions": "other_liabilities_provisions",
    "Total Liabilities": "total_liabilities"
})

# Convert year to integer
financials["year"] = financials["year"].astype(int)

# Create proper financial-year label
financials["financial_year"] = (
    "FY"
    + (financials["year"] - 1).astype(str)
    + "-"
    + financials["year"].astype(str).str[-2:]
)

# Put financial year near the beginning
financials = financials[
    [
        "financial_year",
        "year",
        "bank_name",
        "total_assets",
        "investments",
        "advances",
        "fixed_assets",
        "other_assets",
        "capital",
        "reserves_surplus",
        "deposits",
        "borrowings",
        "other_liabilities_provisions",
        "total_liabilities"
    ]
]

display(financials.head())

print("\nShape:", financials.shape)
print("\nColumns:")
print(financials.columns.tolist())
```
