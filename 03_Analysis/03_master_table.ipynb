```python
import pandas as pd

financials = pd.read_csv("../02_Cleaned_Data/bank_financials.csv")
earnings = pd.read_csv("../02_Cleaned_Data/bank_earnings.csv")
crar = pd.read_csv("../02_Cleaned_Data/bank_crar.csv")
npa = pd.read_csv("../02_Cleaned_Data/bank_npa.csv")
ratios = pd.read_csv("../02_Cleaned_Data/bank_ratios.csv")
```

```python
master=financials.copy()
```

```python
master = master.merge(
    earnings,
    on=["financial_year", "year", "bank_name"],
    how="left"
)
```

```python
master=master.merge(crar,on=["financial_year", "year", "bank_name"],how="left")
```

```python
master=master.merge(npa,on=["financial_year", "year", "bank_name"],how="left")
```

```python
master = master.merge(
    ratios[
        [
            "financial_year",
            "year",
            "bank_name",
            "return_on_assets",
            "return_on_equity",
            "net_npa_to_net_advances"
        ]
    ],
    on=["financial_year", "year", "bank_name"],
    how="left"
)
```
