```python
import pandas as pd

master=pd.read_csv("../02_Cleaned_Data/bank_master.csv")
```

```python
master.head()
```

```python
master.columns.to_list
```

```python
#Finding the total unique banks in the dataset

master['bank_name'].unique().tolist()
```

```python
#Which banks are the largest in India in FY2024-25?

top_banks = (
    master[master["year"] == 2025]
    .sort_values("total_assets", ascending=False)
    [["bank_name", "total_assets"]]
    .head(10)
)

top_banks
```

```python
master = master.sort_values(["bank_name", "year"])
```

```python
master["asset_growth"] = (
    master
    .groupby("bank_name")["total_assets"]
    .pct_change() * 100
)
```

```python
master[master["year"] == 2025].sort_values("asset_growth", ascending=False)[["bank_name", "asset_growth"]].head(10)
```

```python
master["deposit_growth"] = (
    master.groupby("bank_name")["deposits"].pct_change() * 100
)
```

```python
#finding the deposit growth for the year 2025 

master[master["year"]==2025].sort_values("deposit_growth",ascending=False)[["bank_name","deposit_growth"]].head(10)
```

```python
master["advance_growth"]=master.groupby("bank_name")["advances"].pct_change()*100
```

```python
master[[
    "bank_name",
    "year",
    "asset_growth",
    "deposit_growth",
    "advance_growth"
]].head(20)
```

```python
master["calculated_nii"] = (
    master["interest_earned"] - master["interest_expended"]
)

master["nii_difference"] = (
    master["net_interest_income"] - master["calculated_nii"]
)

master["nii_difference"].describe()
```

```python
master[master["year"]==2025].sort_values("net_profit",ascending=False)[["bank_name","net_profit"]].head(10)
```

```python
master[master["year"]==2025].sort_values("return_on_assets",ascending=False)[["bank_name","return_on_assets"]].head(10)
```

```python
#Finding the results for top 5 banks based on assest in the year 2025

master[master["year"]==2025].sort_values("total_assets",ascending=False)[["bank_name","total_assets","net_profit","return_on_assets"]].head(10)
```

```python
master["profit_growth"] = (
    master.groupby("bank_name")["net_profit"].pct_change() * 100
)
```

```python
master.to_csv(
    "../02_Cleaned_Data/bank_master.csv",
    index=False
)
```
